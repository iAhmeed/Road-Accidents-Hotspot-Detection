```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth, association_rules

df_encoded = pd.read_csv("data/processed_accidents_2019.csv")
results_fp = []

min_support_values = [0.01, 0.02, 0.05, 0.10]
max_len_values = [2, 3, 4]

for min_sup in min_support_values:

    # Run FP-Growth ONCE for this support
    itemsets_full = fpgrowth(df_encoded, min_support=min_sup, use_colnames=True)
    itemsets_full["length"] = itemsets_full["itemsets"].apply(len)

    for max_len in max_len_values:
        itemsets = itemsets_full[itemsets_full["length"] <= max_len]

        if len(itemsets) > 0:
            rules = association_rules(itemsets, metric="confidence", min_threshold=0.6)
            num_rules = len(rules)
            avg_conf = rules["confidence"].mean() if num_rules > 0 else 0
            avg_lift = rules["lift"].mean() if num_rules > 0 else 0
        else:
            num_rules = 0
            avg_conf = 0
            avg_lift = 0

        results_fp.append({
            "min_support": min_sup,
            "max_len": max_len,
            "num_itemsets": len(itemsets),
            "num_rules": num_rules,
            "avg_confidence": round(avg_conf, 3),
            "avg_lift": round(avg_lift, 3)
        })

df_results_fp = pd.DataFrame(results_fp)
print(df_results_fp)
```
